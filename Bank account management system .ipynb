# PROJECT 03- BANK ACCOUNT MANAGEMENT SYSTEM
## Suggested Features
* Create Account
* Deposit Money
* Withdraw Money
* Check Balance
* Maintain Transaction History
* Save Data


```python
import json
from datetime import datetime

# We'll store accounts in a list of dictionaries
accounts = []

# -----------------------------
# Core Functions
# -----------------------------

def create_account(name, acc_no, balance=0):
    account = {
        "name": name,
        "acc_no": acc_no,
        "balance": balance,
        "transactions": []
    }
    accounts.append(account)
    print(f" Account created for {name} (Acc No: {acc_no})")

def deposit(acc_no, amount):
    for acc in accounts:
        if acc["acc_no"] == acc_no:
            acc["balance"] += amount
            acc["transactions"].append({"type": "Deposit", "amount": amount, "time": str(datetime.now())})
            print(f" Deposited {amount} into Acc No {acc_no}. New Balance: {acc['balance']}")
            return
    print(" Account not found.")

def withdraw(acc_no, amount):
    for acc in accounts:
        if acc["acc_no"] == acc_no:
            if acc["balance"] >= amount:
                acc["balance"] -= amount
                acc["transactions"].append({"type": "Withdraw", "amount": amount, "time": str(datetime.now())})
                print(f" Withdrawn {amount} from Acc No {acc_no}. New Balance: {acc['balance']}")
            else:
                print(" Insufficient balance.")
            return
    print(" Account not found.")

def check_balance(acc_no):
    for acc in accounts:
        if acc["acc_no"] == acc_no:
            print(f" Balance for Acc No {acc_no}: {acc['balance']}")
            return acc["balance"]
    print(" Account not found.")
    return None

def transaction_history(acc_no):
    for acc in accounts:
        if acc["acc_no"] == acc_no:
            if not acc["transactions"]:
                print("No transactions yet.")
                return
            print(f" Transaction History for Acc No {acc_no}:")
            for t in acc["transactions"]:
                print(f"{t['time']} - {t['type']} {t['amount']}")
            return
    print("Account not found.")

def save_data(filename="accounts.json"):
    with open(filename, "w") as f:
        json.dump(accounts, f)
    print(" Data saved successfully!")

def load_data(filename="accounts.json"):
    global accounts
    try:
        with open(filename, "r") as f:
            accounts = json.load(f)
        print("Data loaded successfully!")
    except FileNotFoundError:
        print(" No saved data found.")

# -----------------------------
# Demo Usage
# -----------------------------
create_account("tej", 1001, 500)
create_account("shriya", 1002, 1000)

deposit(1001, 200)
withdraw(1002, 300)

check_balance(1001)
check_balance(1002)

transaction_history(1001)
transaction_history(1002)

save_data()
load_data()
```
